```python
# renomear todas as colunas do df original "dados"
dados = dados.rename(columns={'Appliances': 'Consumo_Eletrodomesticos',
                              'RH_1': 'RH1',
                              'RH_2': 'RH2',
                              'RH_3': 'RH3'
                              })
dados.head()
```

```python
maior_consumo = dados['Consumo_Eletrodomesticos'].max()
maior_consumo
```

```python
limiar_70 = maior_consumo * 0.7
limiar_70
```

```python
df_filtrado = dados[dados['Consumo_Eletrodomesticos'] >= limiar_70]
df_filtrado.head()
```

```python
total_selecionados = len(df_filtrado)

total_original = len(dados)
percentual = (total_selecionados / total_original) * 100

print(f"Registros selecionados: {total_selecionados}")
print(f"Percentual da amostra: {percentual:.2f}%")
```

```python
# Calcule a temperatura média de T1.
media_T1 = dados['T1'].mean()
print(f" {media_T1:.2f}")
```

```python
# Crie outro DataFrame contendo simultaneamente consumo acima de 70% do máximo e temperatura acima da média.

df_limiar70_e_mediaT1 = dados[
    (dados['Consumo_Eletrodomesticos'] > limiar_70) &
    (dados['T1'] > media_T1)
]
df_limiar70_e_mediaT1.head()
```

```python
# Compare os dois DataFrames e explique o efeito da inclusão da temperatura como segundo critério.
print('A inclusão de temperatura como segundo critério limitou ainda mais os registros acima da limiar 70%,\na condicional "E" tirou mais 1 registro por não estar acima da média de T1.')
```
